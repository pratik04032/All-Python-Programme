```python
# Execute the IPL.ipynb notebook to ensure all necessary files (like pipe.pkl) are generated.
import subprocess

command = [
    'jupyter',
    'nbconvert',
    '--to',
    'notebook',
    '--execute',
    '--ExecutePreprocessor.timeout=600',
    '--ExecutePreprocessor.kernel_name=python3',
    '/content/IPL-WINNING-TEAM-PREDICTION-main/IPL.ipynb',
    '--output',
    '/content/IPL_Final_Execution.ipynb'
]

try:
    subprocess.run(command, check=True)
    print("Notebook executed successfully!")
except subprocess.CalledProcessError as e:
    print(f"Error executing notebook: {e}")
except FileNotFoundError:
    print("Jupyter nbconvert command not found. Please ensure Jupyter is installed.")
```
